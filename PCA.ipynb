```python
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
import matplotlib.pyplot as plt
```

```python
file_path = r'C:\Users\Vaibhav Rathod\Desktop\Nilay\Assignments\Assiggment questions\PCA\wine.csv'
data = pd.read_csv(file_path)
print(data.head())
```

```python
scaler = StandardScaler()
data_standardized = scaler.fit_transform(data)
```

```python
pca = PCA()
pca.fit(data_standardized)
```

```python
explained_variance = pca.explained_variance_ratio_
print(f'Explained variance ratio for each principal component: {explained_variance}')
```

```python
plt.figure(figsize=(8, 6))
plt.plot(range(1, len(explained_variance) + 1), explained_variance, marker='o', linestyle='--')
plt.title('Scree Plot')
plt.xlabel('Principal Components')
plt.ylabel('Explained Variance Ratio')
plt.show()
```

```python
cumulative_variance = pca.explained_variance_ratio_.cumsum()
print(f'Cumulative explained variance: {cumulative_variance}')
```

```python
n_components = next(i for i, cumulative in enumerate(cumulative_variance) if cumulative > 0.95)
print(f'Number of components explaining 95% of the variance: {n_components + 1}')
```

```python
pca = PCA(n_components=n_components + 1)
data_pca = pca.fit_transform(data_standardized)
print(f'Reduced data shape: {data_pca.shape}')
```

```python
if n_components >= 2:
    plt.figure(figsize=(8, 6))
    plt.scatter(data_pca[:, 0], data_pca[:, 1])
    plt.title('PCA - First Two Principal Components')
    plt.xlabel('Principal Component 1')
    plt.ylabel('Principal Component 2')
    plt.show()
```

```python
import pandas as pd
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
import matplotlib.pyplot as plt
```

```python
# Assuming 'X' is the original dataset (2D array or DataFrame)

# Step 1: Standardize the data (optional, but recommended for K-means)
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
# Step 2: Choose the number of clusters (K)
# One common method is the Elbow Method to find the optimal K
# You can run this to visualize the elbow and choose the optimal K

inertia = []  # List to store the inertia values for different K
k_range = range(1, 11)  # Try for K=1 to K=10 clusters
for k in k_range:
    kmeans = KMeans(n_clusters=k, random_state=42)
    kmeans.fit(X_scaled)
    inertia.append(kmeans.inertia_)
```

```python
plt.plot(k_range, inertia, marker='o', linestyle='--')
plt.title('Elbow Method for Optimal K')
plt.xlabel('Number of Clusters (K)')
plt.ylabel('Inertia')
plt.show()
```

```python
kmeans = KMeans(n_clusters=3, random_state=42)  # Replace 3 with your chosen K
kmeans.fit(X_scaled)
```

```python
# Step 4: Inspect the clustering results
# Get the cluster labels assigned to each observation
labels = kmeans.labels_

# Add the cluster labels to the original data (for better interpretation)
X['Cluster'] = labels

# View the first few rows with the cluster labels
print(X.head())
```

```python
print("Cluster Centers:")
print(kmeans.cluster_centers_)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
from mpl_toolkits.mplot3d import Axes3D
from sklearn.decomposition import PCA
```

```python
# Assuming 'X' is the original dataset and 'labels' are the cluster labels obtained from K-means

# 1. 2D Visualization (for 2D data)
if X.shape[1] == 2:  # If the dataset has 2 features
    plt.figure(figsize=(8, 6))
    plt.scatter(X.iloc[:, 0], X.iloc[:, 1], c=labels, cmap='viridis', marker='o')
    plt.title('Clustering Results (2D)')
    plt.xlabel('Feature 1')
    plt.ylabel('Feature 2')
    plt.colorbar(label='Cluster Label')
    plt.show()
# 2. 3D Visualization (for 3D data)
elif X.shape[1] == 3:  # If the dataset has 3 features
    fig = plt.figure(figsize=(8, 6))
    ax = fig.add_subplot(111, projection='3d')
    ax.scatter(X.iloc[:, 0], X.iloc[:, 1], X.iloc[:, 2], c=labels, cmap='viridis')
    ax.set_title('Clustering Results (3D)')
    ax.set_xlabel('Feature 1')
    ax.set_ylabel('Feature 2')
    ax.set_zlabel('Feature 3')
    plt.show()
else:  # For higher-dimensional data (more than 3 features)
    pca = PCA(n_components=2)
    X_pca = pca.fit_transform(X)  # Reduce dimensions to 2

    plt.figure(figsize=(8, 6))
    plt.scatter(X_pca[:, 0], X_pca[:, 1], c=labels, cmap='viridis', marker='o')
    plt.title('Clustering Results (PCA-reduced 2D)')
    plt.xlabel('PC 1')
    plt.ylabel('PC 2')
    plt.colorbar(label='Cluster Label')
    plt.show()
```

```python
from sklearn.decomposition import PCA
from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import silhouette_score
```

```python
# Step 2: Apply PCA
pca = PCA(n_components=2)  # Choose the number of components (e.g., 2 for visualization)
X_pca = pca.fit_transform(X_scaled)
```
