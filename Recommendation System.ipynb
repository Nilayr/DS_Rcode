```python
import pandas as pd
from sklearn.metrics.pairwise import cosine_similarity
from sklearn.preprocessing import MinMaxScaler
import numpy as np

def recommend_anime(df, anime_index, n_recommendations=5):
    """
    Recommend anime based on cosine similarity.
    
    Parameters:
    - df: DataFrame containing the anime dataset
    - anime_index: Index of the anime in the DataFrame for which we want recommendations
    - n_recommendations: Number of recommendations to return
    
    Returns:
    - DataFrame containing the top `n_recommendations` most similar anime
    """
    # 1. Data preprocessing
    # Select relevant columns (ratings, episodes, etc.) and one-hot encode categorical columns
    df_clean = df[['rating', 'episodes', 'popularity']]  # Or add any other features
    df_clean['rating'] = pd.to_numeric(df_clean['rating'], errors='coerce')
    df_clean['episodes'] = pd.to_numeric(df_clean['episodes'], errors='coerce')
    df_clean['popularity'] = pd.to_numeric(df_clean['popularity'], errors='coerce')

    # Fill missing values (using mean of the column for simplicity)
    df_clean['rating'].fillna(df_clean['rating'].mean(), inplace=True)
    df_clean['episodes'].fillna(df_clean['episodes'].mean(), inplace=True)
    df_clean['popularity'].fillna(df_clean['popularity'].mean(), inplace=True)

    # Normalize numerical features
    scaler = MinMaxScaler()
    df_clean[['rating', 'popularity', 'episodes']] = scaler.fit_transform(df_clean[['rating', 'popularity', 'episodes']])

    # 2. Compute cosine similarity matrix
    similarity_matrix = cosine_similarity(df_clean)

    # 3. Get the similarities for the given anime_index
    similarity_scores = similarity_matrix[anime_index]

    # 4. Get indices of top `n_recommendations` excluding the anime itself
    similar_indices = similarity_scores.argsort()[-n_recommendations-1:-1][::-1]
    
    # 5. Return the recommendations
    recommended_anime = df.iloc[similar_indices]
    return recommended_anime

# Example usage (Assuming 'df' i the anime DataFrame with at least 'rating', 'episodes', 'popularity')
# recommended_anime = recommend_anime(df, anime_index=0, n_recommendations=5)
```

```python
import pandas as pd
from sklearn.metrics.pairwise import cosine_similarity
from sklearn.preprocessing import MinMaxScaler

def recommend_anime_with_threshold(df, target_index, similarity_threshold=0.5, n_recommendations=10):
    """
    Recommend anime similar to the target anime based on cosine similarity with a threshold filter.
    
    Parameters:
    - df: DataFrame containing the anime dataset
    - target_index: Index of the target anime in the DataFrame
    - similarity_threshold: Minimum similarity score for recommendations (0 to 1)
    - n_recommendations: Maximum number of recommendations to return
    
    Returns:
    - DataFrame with the most similar anime that pass the similarity threshold
    """
    # Check the size of the DataFrame to avoid out of bounds error
    if target_index >= len(df):
        raise IndexError(f"Target index {target_index} is out of bounds for the DataFrame with {len(df)} rows.")
    
    # 1. Data preprocessing
    # Select relevant columns (e.g., rating, episodes, popularity)
    df_clean = df[['rating', 'episodes', 'popularity']]  # Add other features if needed
    df_clean['rating'] = pd.to_numeric(df_clean['rating'], errors='coerce')
    df_clean['episodes'] = pd.to_numeric(df_clean['episodes'], errors='coerce')
    df_clean['popularity'] = pd.to_numeric(df_clean['popularity'], errors='coerce')

    # Fill missing values with the mean of the column
    df_clean['rating'].fillna(df_clean['rating'].mean(), inplace=True)
    df_clean['episodes'].fillna(df_clean['episodes'].mean(), inplace=True)
    df_clean['popularity'].fillna(df_clean['popularity'].mean(), inplace=True)

    # Normalize numerical features
    scaler = MinMaxScaler()
    df_clean[['rating', 'popularity', 'episodes']] = scaler.fit_transform(df_clean[['rating', 'popularity', 'episodes']])

    # 2. Compute cosine similarity matrix
    similarity_matrix = cosine_similarity(df_clean)

    # 3. Get similarity scores for the target anime
    similarity_scores = similarity_matrix[target_index]

    # 4. Filter out anime below the similarity threshold and create a list of recommendations
    similar_anime = [(index, score) for index, score in enumerate(similarity_scores) if score > similarity_threshold and index != target_index]

    # Sort the recommendations by similarity score in descending order
    similar_anime = sorted(similar_anime, key=lambda x: x[1], reverse=True)

    # 5. Limit to `n_recommendations` and return the result
    top_recommendations = similar_anime[:n_recommendations]

    # Get the indices of the recommended anime
    recommended_indices = [index for index, _ in top_recommendations]

    # Return the recommended anime from the DataFrame
    recommended_anime = df.iloc[recommended_indices]
    
    return recommended_anime

# Example usage:
# Assuming `df` is the anime DataFrame with at least 3 rows
# Adjust similarity threshold to experiment with different recommendation sizes
try:
    recommended_anime = recommend_anime_with_threshold(df, target_index=5, similarity_threshold=0.7, n_recommendations=5)
    print(recommended_anime[['name', 'rating', 'episodes']])
except IndexError as e:
    print(e)
```

```python
def recommend_anime_with_threshold(df, target_index, similarity_threshold=0.7, n_recommendations=5):
    # Ensure the target_index is valid
    if target_index >= len(df) or target_index < 0:
        raise IndexError(f"Target index {target_index} is out of bounds for the DataFrame with {len(df)} rows.")
    
    # 1. Data preprocessing (select relevant columns)
    df_clean = df[['rating', 'episodes']].dropna()  # Remove rows with NaN values
    
    # 2. Normalize numerical features
    df_clean['rating'] = pd.to_numeric(df_clean['rating'], errors='coerce')
    df_clean['episodes'] = pd.to_numeric(df_clean['episodes'], errors='coerce')
    
    scaler = MinMaxScaler()
    df_clean[['rating', 'episodes']] = scaler.fit_transform(df_clean[['rating', 'episodes']])
    
    # 3. Calculate cosine similarity between anime
    similarity_matrix = cosine_similarity(df_clean)
    
    # 4. Get similarity scores for the target anime
    similarity_scores = similarity_matrix[target_index]
    
    # 5. Filter out anime below the similarity threshold
    similar_anime = [(index, score) for index, score in enumerate(similarity_scores) if score > similarity_threshold and index != target_index]
    
    # Sort by similarity score in descending order
    similar_anime = sorted(similar_anime, key=lambda x: x[1], reverse=True)
    
    # 6. Get top recommendations
    recommended_anime = [df.iloc[index] for index, _ in similar_anime[:n_recommendations]]
    
    return pd.DataFrame(recommended_anime)

# Example usage:
recommended_anime = recommend_anime_with_threshold(df, target_index=0, similarity_threshold=0.7, n_recommendations=5)
print(recommended_anime[['name', 'rating', 'episodes']])
```

```python
from sklearn.metrics import precision_score, recall_score, f1_score

def evaluate_recommendation_system(user_item_matrix, get_top_n_recommendations, top_n=10):
    true_positives = []
    false_positives = []
    false_negatives = []
    
    for user_id in user_item_matrix.index:
        # Actual interactions for the user
        actual_items = user_item_matrix.loc[user_id]
        actual_items = actual_items[actual_items > 0].index.tolist()  # Non-zero interactions as actual items
        
        # Predicted recommendations for the user
        recommended_items = get_top_n_recommendations(user_id, top_n)
        
        # Calculate True Positives, False Positives, and False Negatives
        for item in recommended_items:
            if item in actual_items:
                true_positives.append(1)
            else:
                false_positives.append(1)
        
        for item in actual_items:
            if item not in recommended_items:
                false_negatives.append(1)

    # Precision, Recall, F1 Score
    precision = sum(true_positives) / (sum(true_positives) + sum(false_positives)) if sum(true_positives) + sum(false_positives) > 0 else 0
    recall = sum(true_positives) / (sum(true_positives) + sum(false_negatives)) if sum(true_positives) + sum(false_negatives) > 0 else 0
    f1 = 2 * precision * recall / (precision + recall) if precision + recall > 0 else 0

    return precision, recall, f1
```
